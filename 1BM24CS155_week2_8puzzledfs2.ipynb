```python
import copy

class PuzzleNode:
    def __init__(self, board, parent=None, move="", depth=0):
        self.board = board
        self.parent = parent
        self.move = move
        self.depth = depth

    def get_blank_position(self):
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return r, c

    def get_neighbors(self):
        neighbors = []

        r, c = self.get_blank_position()

        moves = [
            (-1, 0, "Up"),
            (1, 0, "Down"),
            (0, -1, "Left"),
            (0, 1, "Right")
        ]

        for dr, dc, move_name in moves:
            nr, nc = r + dr, c + dc

            if 0 <= nr < 3 and 0 <= nc < 3:

                new_board = copy.deepcopy(self.board)

                new_board[r][c], new_board[nr][nc] = \
                    new_board[nr][nc], new_board[r][c]

                neighbors.append(
                    PuzzleNode(
                        new_board,
                        self,
                        move_name,
                        self.depth + 1
                    )
                )

        return neighbors


# DFS
def solve_dfs(start_board, goal_board):

    start_node = PuzzleNode(start_board)

    goal_tuple = tuple(tuple(row) for row in goal_board)

    # Stack for DFS
    stack = [start_node]

    # Visited states
    visited = set()

    while stack:

        # Remove top node
        current_node = stack.pop()

        current_tuple = tuple(
            tuple(row) for row in current_node.board
        )

        # Check goal
        if current_tuple == goal_tuple:
            return current_node

        # Skip if already visited
        if current_tuple in visited:
            continue

        visited.add(current_tuple)

        # Add neighbors to stack
        for neighbor in current_node.get_neighbors():

            neighbor_tuple = tuple(
                tuple(row) for row in neighbor.board
            )

            if neighbor_tuple not in visited:
                stack.append(neighbor)

    return None


# Input function
def parse_input(prompt_text):

    print(prompt_text)

    board = []

    for _ in range(3):
        row = list(map(int, input().split()))
        board.append(row)

    return board


# Main program
if __name__ == "__main__":

    initial_board = parse_input(
        "Enter Initial State (3 rows of 3 space-separated numbers):"
    )

    goal_board = parse_input(
        "\nEnter Goal State (3 rows of 3 space-separated numbers):"
    )

    solution_node = solve_dfs(
        initial_board,
        goal_board
    )

    if solution_node:

        # Only number of moves
        print("\nNumber of Moves:", solution_node.depth)

        # Complexity
        print("\n--- Complexity ---")
        print("Time Complexity: O(b^m)")
        print("Space Complexity: O(b^m)")
        print("For 8-Puzzle, branching factor b <= 4.")

    else:

        print("\nNo solution found.")

        print("\n--- Complexity ---")
        print("Time Complexity: O(b^m)")
        print("Space Complexity: O(b^m)")
        print("For 8-Puzzle, branching factor b <= 4.")
```
