```python

import random
import math

def cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts

def display(board):
    for r in range(8):
        for c in range(8):
            if board[c] == r:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

def simulated_annealing():
    n = 8
    board = [random.randrange(n) for _ in range(n)]
    T = 10

    for step in range(10000):
        current_cost = cost(board)

        if current_cost == 0:
            return board, current_cost, step + 1

        new_board = board.copy()
        col = random.randrange(n)
        new_board[col] = random.randrange(n)

        new_cost = cost(new_board)
        delta = new_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            board = new_board

        T *= 0.99

        if T < 0.001:
            break

    return board, cost(board), step + 1

board, result, steps = simulated_annealing()

print("Final Board:", board)
print("Final Cost:", result)
print("Iterations:", steps)

if result == 0:
    print("Solution found!")
    display(board)
else:
    print("No solution found in this run.")
```
