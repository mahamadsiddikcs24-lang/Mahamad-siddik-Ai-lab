```python
import heapq

def calculate_hn(state, goal):
    hn = 0
    goal_pos = {val: (r, c) for r, row in enumerate(goal) for c, val in enumerate(row)}
    for r in range(3):
        for c in range(3):
            val = state[r][c]
            if val != 0:
                gr, gc = goal_pos[val]
                hn += abs(r - gr) + abs(c - gc)
    return hn

def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    neighbors = []
    r, c = find_blank(state)
    moves = [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]
    for dr, dc, move_name in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            neighbors.append((tuple(tuple(row) for row in new_state), move_name))
    return neighbors

def print_board(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()

def solve_8_puzzle(start, goal):
    start_t = tuple(tuple(row) for row in start)
    goal_t = tuple(tuple(row) for row in goal)

    open_list = []
    counter = 0
    h0 = calculate_hn(start_t, goal_t)
    heapq.heappush(open_list, (0 + h0, 0, counter, start_t, []))
    closed_list = set()

    print("Initial State:")
    print(f"g(n) = 0, h(n) = {h0}, f(n) = {h0}")
    print_board(start_t)

    while open_list:
        f, g, _, current, path = heapq.heappop(open_list)

        if current == goal_t:
            return path

        if current in closed_list:
            continue
        closed_list.add(current)

        for neighbor, move in get_neighbors(current):
            if neighbor not in closed_list:
                gn = g + 1
                hn = calculate_hn(neighbor, goal_t)
                fn = gn + hn
                counter += 1
                heapq.heappush(open_list, (fn, gn, counter, neighbor, path + [(move, neighbor, gn, hn, fn)]))

if __name__ == "__main__":
    # Matrices matched exactly to your image input (0 = blank space)
    initial_board = [
        [2, 8, 3],
        [1, 6, 4],
        [7, 0, 5]
    ]

    goal_board = [
        [1, 2, 3],
        [8, 0, 4],
        [7, 6, 5]
    ]

    solution_path = solve_8_puzzle(initial_board, goal_board)

    if solution_path:
        for step, (move, state, gn, hn, fn) in enumerate(solution_path, 1):
            print(f"Step {step}: Move {move}")
            print(f"g(n) = {gn}, h(n) = {hn}, f(n) = {fn}")
            print_board(state)
    else:
        print("No solution found for this configuration.")
```
