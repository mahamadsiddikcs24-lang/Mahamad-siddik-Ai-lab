```python
def calculate_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def print_diagram(state):
    n = len(state)
    print("+" + "---+" * n)
    for row in range(n):
        row_str = "|"
        for col in range(n):
            if state[col] == row:
                row_str += " Q |"
            else:
                row_str += "   |"
        print(row_str)
        print("+" + "---+" * n)

def hill_climbing_nqueens(initial_state):
    current_state = list(initial_state)
    n = len(current_state)

    while True:
        current_cost = calculate_conflicts(current_state)
        state_str = ", ".join(f"x{i} = {current_state[i]}" for i in range(n))
        print(f"Neighbours of {state_str} (cost is {current_cost})")

        if current_cost == 0:
            print("\nGoal configuration reached successfully:")
            print_diagram(current_state)
            break

        best_neighbor = None
        best_cost = current_cost
        chosen_pair = None
        neighbors_list = []

        for i in range(n):
            for j in range(i + 1, n):
                neighbor = list(current_state)
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
                cost = calculate_conflicts(neighbor)
                neighbors_list.append(((i, j), neighbor, cost))

        for (i, j), neighbor, cost in neighbors_list:
            if cost < best_cost:
                best_cost = cost
                best_neighbor = neighbor
                chosen_pair = (i, j)

        for (i, j), neighbor, cost in neighbors_list:
            neighbor_str = ", ".join(f"x{k} = {neighbor[k]}" for k in range(n))
            if chosen_pair and (i, j) == chosen_pair:
                print(f" • {neighbor_str} cost is {cost}. Choose this neighbour.")
            else:
                print(f" • {neighbor_str} cost is {cost}.")

        if best_neighbor is None or best_cost >= current_cost:
            print("Terminated: Stuck in a local minimum.")
            break

        current_state = best_neighbor
        print()

if __name__ == "__main__":
    initial_state = [3, 1, 2, 0]
    hill_climbing_nqueens(initial_state)
```
