# Agentic Tic-Tac-Toe Game

First, let's define the game board and basic game functions.

```python
import random
print("name:siddik")
print("usn:1BM24CS155")
def create_board():
    return [' ' for _ in range(9)]

def display_board(board):
    for i in range(0, 9, 3):
        print(f'{board[i]}|{board[i+1]}|{board[i+2]}')
        if i < 6:
            print('-+-+-')

def is_valid_move(board, move):
    return 0 <= move < 9 and board[move] == ' '

def make_move(board, move, player):
    board[move] = player

def check_win(board, player):
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8], # Rows
        [0, 3, 6], [1, 4, 7], [2, 5, 8], # Columns
        [0, 4, 8], [2, 4, 6]            # Diagonals
    ]
    for condition in win_conditions:
        if all(board[i] == player for i in condition):
            return True
    return False

def check_draw(board):
    return ' ' not in board
def get_random_agent_move(board):
    available_moves = [i for i, spot in enumerate(board) if spot == ' ']
    if available_moves:
        return random.choice(available_moves)
    return -1 # No available moves
def play_game():
    board = create_board()
    current_player = 'X' # Human player
    agent_player = 'O'   # AI agent

    print("Welcome to Tic-Tac-Toe! You are 'X'.")
    display_board(board)

    while True:
        if current_player == 'X':
            try:
                move = int(input("Enter your move (0-8): "))
                if not is_valid_move(board, move):
                    print("Invalid move. Try again.")
                    continue
            except ValueError:
                print("Invalid input. Please enter a number between 0 and 8.")
                continue
        else:
            print(f"Agent ({agent_player}) is making a move...")
            move = get_random_agent_move(board)
            if move == -1:
                print("No moves left for agent.")
                break

        make_move(board, move, current_player)
        display_board(board)

        if check_win(board, current_player):
            print(f"Player {current_player} wins!")
            break
        elif check_draw(board):
            print("It's a draw!")
            break

        current_player = 'O' if current_player == 'X' else 'X'

play_game()
```
