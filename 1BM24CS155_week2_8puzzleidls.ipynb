```python
import copy

class PuzzleNode:
    def __init__(self, board, parent=None, move="", depth=0):
        self.board = board
        self.parent = parent
        self.move = move
        self.depth = depth

    def get_blank_position(self):
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return r, c

    def get_neighbors(self):
        neighbors = []
        r, c = self.get_blank_position()
        moves = [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]

        for dr, dc, move_name in moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board = copy.deepcopy(self.board)
                new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
                neighbors.append(PuzzleNode(new_board, self, move_name, self.depth + 1))

        return neighbors

def dls(current_node, goal_tuple, limit, visited):
    board_tuple = tuple(tuple(row) for row in current_node.board)

    if board_tuple == goal_tuple:
        return current_node

    if current_node.depth >= limit:
        return None

    visited.add(board_tuple)

    for neighbor in current_node.get_neighbors():
        neighbor_tuple = tuple(tuple(row) for row in neighbor.board)
        if neighbor_tuple not in visited:
            result = dls(neighbor, goal_tuple, limit, visited)
            if result:
                return result

    visited.remove(board_tuple)
    return None

def solve_iddfs(start_board, goal_board):
    start_node = PuzzleNode(start_board)
    goal_tuple = tuple(tuple(row) for row in goal_board)

    for limit in range(50):
        visited = set()
        result = dls(start_node, goal_tuple, limit, visited)
        if result:
            return result
    return None

def print_moves_only(node):
    path = []
    current = node
    while current:
        if current.move:
            path.append(current.move)
        current = current.parent
    path.reverse()

    print(f"\nTotal Moves: {len(path)}")
    print("Path: " + " -> ".join(path))

def parse_input(prompt_text):
    print(prompt_text)
    board = []
    for _ in range(3):
        row = list(map(int, input().split()))
        board.append(row)
    return board

if __name__ == "__main__":
    initial_board = parse_input("Enter Initial State (3 rows of 3 space-separated numbers):")
    goal_board = parse_input("\nEnter Goal State (3 rows of 3 space-separated numbers):")

    solution_node = solve_iddfs(initial_board, goal_board)

    if solution_node:
        print_moves_only(solution_node)
    else:
        print("\nNo solution found within depth limits.")
```
